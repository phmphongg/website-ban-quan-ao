### 20. (6đ) Tỉ lệ người married hay single men (marital-status feature) chiếm nhiều hơn trong số những người kiếm được >50k? Những người có tình trạng hôn nhân (married) có marital-status feature bắt đầu với Married (Married-civ-spouse, Married-spouse-absent or Married-AF-spouse), còn lại được coi là single men.

```python
# You code here
```

```python
rich_people = df[df['salary']=='>50K']

married = rich_people[
    rich_people['marital-status'].str.startswith('Married')
]

single = rich_people[
    ~rich_people['marital-status'].str.startswith('Married')
]

print("Married:", len(married))
print("Single:", len(single))
```

```python

if len(married) > len(single):
    print("Married nhiều hơn")
else:
    print("Single nhiều hơn")
```

### 21. (6đ) Số giờ tối đa một người làm việc mỗi tuần (hours-per-week feature) là bao nhiêu? Có bao nhiêu người làm việc với số giờ như vậy và tỷ lệ những người kiếm được nhiều (>50K) trong số đó là bao nhiêu?

```python
# You code here
```

```python
max_hours = df['hours-per-week'].max()

workers = df[df['hours-per-week']==max_hours]

num_workers = len(workers)

rich_ratio = (
    (workers['salary']=='>50K').sum()
    / num_workers
    * 100
)

print("Max hours:", max_hours)
print("Number workers:", num_workers)
print("Rich ratio:", rich_ratio)
```

### 23. (10đ) Số người da trắng (white in race feature) kiếm được số tiền >50k ở nước Mỹ chiếm bao nhiêu phần trăm. Câu hỏi tương tự với người da đen (black in race feature).

```python
# You code here
usa = df[df['native-country'] == 'United-States']

white_usa = usa[usa['race'] == 'White']

white_percent = (
    (white_usa['salary'] == '>50K').sum()
    / len(white_usa)
    * 100
)

black_usa = usa[usa['race'] == 'Black']

black_percent = (
    (black_usa['salary'] == '>50K').sum()
    / len(black_usa)
    * 100
)

print("Tỷ lệ người da trắng ở Mỹ kiếm >50K: {:.2f}%".format(white_percent))
print("Tỷ lệ người da đen ở Mỹ kiếm >50K: {:.2f}%".format(black_percent))
```
